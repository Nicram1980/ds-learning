## Zadanie 1 – Kalkulator BMI
* Napisz program, który obliczy wskaźnik BMI (Body Mass Index).
### Dane:
* Waga: 70 kg
* Wzrost: 1.75 m
### Wymagania:
#### Stwórz zmienne
* waga
* wzrost
* Oblicz BMI według wzoru: BMI = waga / (wzrost²)
* Wyświetl wynik z 2 miejscami po przecinku
* Dodaj interpretację: BMI < 18.5 (niedowaga), 18.5-24.9 (norma), ≥25 (nadwaga)

```python
waga = 70
wzrost = 1.75

# z podanych danych obliczamy BMI według wzoru i wyświetlamy wynik z 2 miejscami po przecinku oraz interpretację
bmi = waga / (wzrost ** 2)
print(f"BMI: {bmi:.2f}")

# wyswietlam interpretację BMI w zależności od wartości
if bmi < 18.5:
    print("Interpretacja: niedowaga")
elif 18.5 <= bmi < 25:
    print("Interpretacja: norma")
else:
    print("Interpretacja: nadwaga")

```

## Zadanie 2 – Konwersja temperatur
## Napisz program konwertujący temperatury Celsius na Fahrenheit i Kelvin.
### Dane:
* Temperatura: 23.5°C
### Wymagania:
* Fahrenheit = Celsius × 9/5 + 32
* Kelvin = Celsius + 273.15
* Wyświetl wszystkie 3 wartości sformatowane z 1 miejscem po przecinku

```python
temperatura_celsius = 23.5

# zamieniamy temperaturę w Celsius na Fahrenheit i Kelvin korzystając z podanych wzorów
fahrenheit = temperatura_celsius * 9/5 + 32
kelvin = temperatura_celsius + 273.15

print(f"Temperatura w Celsius: {temperatura_celsius:.1f}°C")
print(f"Temperatura w Fahrenheit: {fahrenheit:.1f}°F")
print(f"Temperatura w Kelvin: {kelvin:.1f}K")
```

## Zadanie 11 – Sprawdzanie poprawności hiperparametrów
### Zwaliduj hiperparametry przed treningiem.
### Dane:
* learning_rate = 0.01
* batch_size = 64
* epochs = 100
* dropout = 0.5
### Wymagania:
* Sprawdź czy learning_rate w zakresie (0, 1]
* Sprawdź czy batch_size jest potęgą 2 (użyj operacji bitowych lub logarytmów)
* Sprawdź czy epochs > 0 i < 1000
* Sprawdź czy dropout w zakresie [0, 1]
* Wyświetl "OK" lub komunikat błędu dla każdego parametru

```python
import math
learning_rate = 0.01
batch_size = 64
epochs = 100
dropout = 0.5
```

```python
# spawdzamy czy learning_rate w zakresie (0, 1]
learning_rate_ok = 0 < learning_rate <= 1

# sprawdzamy czy batch_size jest potęgą 2
batch_size_ok = batch_size > 0 and (math.log2(batch_size) % 1 == 0)

# sprawdzamy czy epochs > 0 i < 1000
epochs_ok = 0 < epochs < 1000

# sprawdzamy czy dropout w zakresie [0, 1]
dropout_ok = 0 <= dropout <= 1
```

```python
print(f"learning_rate: {'OK' if learning_rate_ok else 'Error: learning_rate must be in (0, 1]'}")
print(f"batch_size: {'OK' if batch_size_ok else 'Error: batch_size must be a power of 2'}")
print(f"epochs: {'OK' if epochs_ok else 'Error: epochs must be > 0 and < 1000'}")
print(f"dropout: {'OK' if dropout_ok else 'Error: dropout must be in [0, 1]'}")
```

## Zadanie 12 – Estymacja czasu treningu
### Oblicz szacowany czas treningu modelu.
### Dane:
* Liczba próbek: 50000
* Batch size: 128
* Liczba epok: 50
* Czas na jeden batch: 0.15 sekundy
## Wymagania:
* Oblicz liczbę batchy na epokę (użyj dzielenia całkowitego i modulo)
* Oblicz całkowity czas w sekundach
* Skonwertuj na minuty i godziny
* Wyświetl: "Szacowany czas: X h Y min Z s"




```python
liczba_probek = 50000
batch_size = 128
liczba_epok = 50
czas_na_jeden_batch = 0.15
```

```python
# liczymy ile batchy wejdzie na epokę
liczba_batchy_na_epoke = liczba_probek // batch_size
liczba_batchy_modulo = liczba_probek % batch_size
całkowita_liczba_batchy = liczba_batchy_na_epoke + (1 if liczba_batchy_modulo > 0 else 0)

print(f"Liczba batchy na epokę: {liczba_batchy_na_epoke}")
print(f"całkowita liczba batchy:{całkowita_liczba_batchy}")

# liczymy łączną liczbę epok
laczna_liczba_epok = liczba_epok * całkowita_liczba_batchy

print(f"Łączna liczba batchy: {laczna_liczba_epok}")

# liczymy łączny czas w sekundach

całkowity_czas_w_sekundach = laczna_liczba_epok * czas_na_jeden_batch
print(f"Całkowity czas w sekundach: {całkowity_czas_w_sekundach:.2f} s")

# konwertujemy czas na sekundy, minuty i godziny
czas_w_godzinach = int(całkowity_czas_w_sekundach // 3600)
czas_w_minutach = int((całkowity_czas_w_sekundach % 3600) // 60)
czs_w_sekundach = int(całkowity_czas_w_sekundach // 60)

# Wyświetl: "Szacowany czas: X h Y min Z s"
szacowany_czas = f"Szacowany czas: {czas_w_godzinach} h {czas_w_minutach} min {int(całkowity_czas_w_sekundach % 60)} s"
print(szacowany_czas)
```

## Zadanie 15 – Standaryzacja Z-score
### Standaryzuj wartości używając wzoru z-score.
### Dane (wagi pacjentów w kg):
* 65.5, 78.2, 54.8, 92.1, 70.0
### Wymagania:
* Oblicz średnią i odchylenie standardowe
* Dla każdej wartości oblicz z-score: z = (x - mean) / std
* Sprawdź, że średnia z-scores ≈ 0 i odchylenie ≈ 1
* Wyświetl oryginalną wartość i jej z-score
* Zinterpretuj: z > 2 (wartość bardzo wysoka), |z| < 1 (typowa), z < -2 (bardzo niska)

```python
wagi_pacjentow = [65.5, 78.2, 54.8, 92.1, 70.0]

# obliczam średnią i odchylenie standardowe dla wag pacjentów
srednia = sum(wagi_pacjentow) / len(wagi_pacjentow)

waga_pacjent_1 = wagi_pacjentow[0]
waga_pacjent_2 = wagi_pacjentow[1]
waga_pacjent_3 = wagi_pacjentow[2]
waga_pacjent_4 = wagi_pacjentow[3]
waga_pacjent_5 = wagi_pacjentow[4]

kwadrat_roznica_p_1 = (waga_pacjent_1 - srednia) ** 2
kwadrat_roznica_p_2 = (waga_pacjent_2 - srednia) ** 2
kwadrat_roznica_p_3 = (waga_pacjent_3 - srednia) ** 2
kwadrat_roznica_p_4 = (waga_pacjent_4 - srednia) ** 2
kwadrat_roznica_p_5 = (waga_pacjent_5 - srednia) ** 2

wariacja = (kwadrat_roznica_p_1 + kwadrat_roznica_p_2 + kwadrat_roznica_p_3 + kwadrat_roznica_p_4 + kwadrat_roznica_p_5) / len(wagi_pacjentow)

std = wariacja ** 0.5

# obliczam z-score dla każdej wartości wagi pacjenta
z_scores_p_1 = (waga_pacjent_1 - srednia) / std
z_scores_p_2 = (waga_pacjent_2 - srednia) / std
z_scores_p_3 = (waga_pacjent_3 - srednia) / std
z_scores_p_4 = (waga_pacjent_4 - srednia) / std
z_scores_p_5 = (waga_pacjent_5 - srednia) / std

# sprawdzam czy srednia z-score jest bliska 0 i odchylenie z-score bliskie 1
srednia_z = (z_scores_p_1 + z_scores_p_2 + z_scores_p_3 + z_scores_p_4 + z_scores_p_5) / 5

# obliczam odchylenie z z-score dla każdego pacjenta
odchyleni_p_1 = (z_scores_p_1 - srednia_z) ** 2
odchyleni_p_2 = (z_scores_p_2 - srednia_z) ** 2
odchyleni_p_3 = (z_scores_p_3 - srednia_z) ** 2
odchyleni_p_4 = (z_scores_p_4 - srednia_z) ** 2
odchyleni_p_5 = (z_scores_p_5 - srednia_z) ** 2

warjacja_z = (odchyleni_p_1 + odchyleni_p_2 + odchyleni_p_3 + odchyleni_p_4 + odchyleni_p_5)/5
```

```python
print(f"Średnia: {srednia:.2f}")
print(f"Wariacja: {wariacja:.2f}")
print(f"Odchylenie standardowe: {std:.2f}")
print(f"z scores pacjęta 1: {z_scores_p_1:.2f} normalizacja: {'wartość bardzo wysoka' if z_scores_p_1 > 2 else 'bardzo niska' if z_scores_p_1 < -2 else 'typowa' if abs(z_scores_p_1) < 1 else 'poza zakresem zadania'}")
print(f"z scores pacjęta 2: {z_scores_p_2:.2f} normalizacja: {'wartość bardzo wysoka' if z_scores_p_2 > 2 else 'bardzo niska' if z_scores_p_2 < -2 else 'typowa' if abs(z_scores_p_2) < 1 else 'poza zakresem zadania'}")
print(f"z scores pacjęta 3: {z_scores_p_3:.2f} normalizacja: {'wartość bardzo wysoka' if z_scores_p_3 > 2 else 'bardzo niska' if z_scores_p_3 < -2 else 'typowa' if abs(z_scores_p_3) < 1 else 'poza zakresem zadania'}")
print(f"z scores pacjęta 4: {z_scores_p_4:.2f} normalizacja: {'wartość bardzo wysoka' if z_scores_p_4 > 2 else 'bardzo niska' if z_scores_p_4 < -2 else 'typowa' if abs(z_scores_p_4) < 1 else 'poza zakresem zadania'}")
print(f"z scores pacjęta 5: {z_scores_p_5:.2f} normalizacja: {'wartość bardzo wysoka' if z_scores_p_5 > 2 else 'bardzo niska' if z_scores_p_5 < -2 else 'typowa' if abs(z_scores_p_5) < 1 else 'poza zakresem zadania'}")

print((f"Srednia z s-score: {abs(round(srednia_z))}"))
print(f"Odchylenie z s-score: {abs(round(warjacja_z,2))}")
```
