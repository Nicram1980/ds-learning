## Zadanie 1 – Tworzenie i indeksowanie listy
### Stwórz listę zawierającą pomiary temperatury z 7 dni:
* [22.5, 23.0, 21.5, 24.0, 23.5, 22.0, 23.0]
### Wyświetl:
* temperaturę z pierwszego dnia,
* temperaturę z ostatniego dnia,
* temperatury z trzech środkowych dni (slicing).

```python
temperatura_lista = [22.5, 23.0, 21.5, 24.0, 23.5, 22.0, 23.0]
temperatura_dzien_pierwszy = temperatura_lista[0]
temperatura_dzien_ostatni = temperatura_lista[-1]
temperatura_z_trzech_dni = temperatura_lista[2:4]
```

```python
print(temperatura_dzien_pierwszy)
print(temperatura_dzien_ostatni)
print(temperatura_z_trzech_dni)
temperatura_len = len(temperatura_lista)
print(temperatura_len)
temperatura_sum = sum(temperatura_lista)
temperatura_std = (temperatura_sum / temperatura_len) ** 0.5
print(temperatura_std)
```
